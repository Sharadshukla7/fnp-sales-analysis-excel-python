# total revenue

```python
total_sales = df['total_sales'].sum()
```

```python
print("total revenue : ", total_sales)
```

# average customer spent

```python
avg_cust_spent = df['total_sales'].mean()
```

```python
print("Average customer spent : ", round(avg_cust_spent,2))
```

# top 5 products

```python
top5products = df.groupby('Product_Name')['total_sales'].sum().sort_values(ascending=False).head(5)
```

```python
print("top 5 products : ",top5products)
```

# monthly revenue

```python
monthely_revenue = df.groupby('month_name')['total_sales'].sum()
```

```python
print("monthly revenue : ", monthely_revenue)
```

```python
df['delivery_date'] = pd.to_datetime(df['delivery_date'])
```

```python
df['hour'] = df['delivery_date'].dt.hour
```

# top categories

```python
top7category= df.groupby('category')['total_sales'].sum().sort_values(ascending=False).head(7)
```

```python
print(" top 7 category : ", top7category)
```

# average order delivery time 

ave_order_delivery_time = df['diffrence_order_delivery'].mean()

```python
print("average order delivery time : ", ave_order_delivery_time)
```

# top 10 cities by number of order

```python
top10cities_by_number_of_order = df.groupby('city')['order_id'].count().sort_values(ascending=False).head(10)
```

```python
print (" top 10 cities by number of order : ", top10cities_by_number_of_order)
```

# revenue comparison between occassions

```python
revenue_comparison_bt_occassions = df.groupby('occasion')['total_sales'].sum().sort_values(ascending=False)
```

```python
print("revenue comparison between occassions : " , revenue_comparison_bt_occassions)
```

# Product Popularity By Occasion

```python
product_popularity_by_occassions = df.groupby(['occasion', 'product_name'])['total_sales'].sum().reset_index().sort_values(['occasion', 'total_sales'],ascending=[True, False])
```

```python
print("Product Popularity By Occasion : ", product_popularity_by_occassions)
```

```python
populariti_by_occasions = df.groupby(['occasion','product_name'])['total_sales'].sum().reset_index().sort_values(['occasion', 'total_sales'],ascending=[True,False]) 
```

```python
print("The Popularity Of The Products By Occasions Are : ", populariti_by_occasions)
```
